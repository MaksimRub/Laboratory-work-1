```python
import requests
import re
from html.parser import HTMLParser
class MyFile:
    file_name = ""
    mode = ""
    def __init__(self, file_name, mode):
        self.file_name = file_name
        self.mode = mode
    def write(self,text):
        try:
            if self.mode == "write":
                file = open (self.file_name, "w")
                file.write(text)
                file.close()
                return "файл записан успешно"
            elif self.mode == "append":
                file = open (self.file_name, "a")
                file.write(text)
                file.close()
                return "файл успешно дополнен"
            else:
                print("неверно выбран режим работы с файлом")
        except Exception as ex:
            return "при записи в файл возникла ошибка: " + str(ex)
    def read(self):
        try:
            if self.mode == "read":
                file = open (self.file_name, "r")
                text = file.read()
                file.close()
                return text
            else:
                print("неверно выбран режим работы с файлом")
        except Exception as ex:
            return "при чтении из файла возникла ошибка: " + str(ex)
        
    def get_text_url(self, text):
        class MyHTMLParser(HTMLParser):
            def __init__(self):
                super().__init__()
                self.text = []

            def handle_data(self, data):
                if data.strip():  # удаляем пустые строки
                    self.text.append(data.strip())
        parser = MyHTMLParser()
        parser.feed(text)
        text1 = '\n'.join(parser.text)
        return text1

    
    def read_url(self):
        try:
            if self.mode == "url":
                web_page = requests.get(self.file_name)
                text = self.get_text_url(web_page.text)
                return text
            else:
                print("неверно выбран режим работы с файлом")
        except Exception as ex:
            return "при чтении из файла возникла ошибка: " + str(ex)
    def count_urls(self):
        try:
            if self.mode == "url":
                file = requests.get(self.file_name)
                text = file.text
                reg = r'https?://[^\s]+'
                return len(re.findall(reg,text))
            else:
                print("неверно выбран режим работы с файлом")
        except Exception as ex:
            return "при чтении из файла возникла ошибка: " + str(ex)
        
    def write_url(self,destination):
        try:
            if self.mode == "url":
                web_page = requests.get(self.file_name)
                text = self.get_text_url(web_page.text)
                file = open(destination,"w")
                file.write(text)
                file.close()
                return "файл записан успешно"
            else:
                print("неверно выбран режим работы с файлом")
        
        
        except Exception as ex:
            return "при записи в файл возникла ошибка: " + str(ex)

        
```
